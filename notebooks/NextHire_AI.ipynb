# 📌 Library Verification

## Objective

Before starting development, we must verify that all required Python libraries are installed correctly.

Libraries Used:

- Pandas
- NumPy
- Matplotlib
- ReportLab
- PyMuPDF
- Transformers
- PyTorch

```python
# ==========================================
# NextHire AI - Library Verification
# ==========================================

libraries = {
    "Pandas": "pandas",
    "NumPy": "numpy",
    "Matplotlib": "matplotlib",
    "ReportLab": "reportlab",
    "PyMuPDF": "fitz",
    "Transformers": "transformers",
    "PyTorch": "torch"
}

print("=" * 50)
print("📦 Library Verification")
print("=" * 50)

for name, module in libraries.items():
    try:
        imported_module = __import__(module)

        # Get version if available
        version = getattr(imported_module, "__version__", "Version Not Available")

        print(f"✅ {name:<15} | Installed | Version: {version}")

    except ImportError:
        print(f"❌ {name:<15} | NOT Installed")

print("\n🎉 Verification Completed!")
```
